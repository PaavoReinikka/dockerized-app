```python
import psycopg

try:
    conn = psycopg.connect(
        host="localhost", #Note: not the service name when connecting from the host
        port=5432, # default PostgreSQL port
        user="username",      # (project.env)
        password="password",  # (project.env)
        dbname="database"     # (project.env)
    )
    print("Connection successful!")
except Exception as e:
    print("Connection failed:", e)
```

```python
# Test the connection by fetching the PostgreSQL version
try:
    with conn.cursor() as cur:
        cur.execute("SELECT version();")
        version = cur.fetchone()
        print("PostgreSQL version:", version[0])
except Exception as e:
    print("Connection test failed:", e)
```

```python
conn.close()
```
